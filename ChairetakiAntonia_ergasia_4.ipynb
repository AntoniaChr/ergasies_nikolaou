Exercise IV. Burrows-Wheeler Transformation

The following code decoding the BWT received from my colleague Thodoris Gionnas, and after I search the produced protein sequence in UniProt database I end up that the original protein sequence was from Dehydrogenase eriK protein (ERIK_HERER), with UniProt ID: P0DO29

```python
def inverse_bwt(bwt):
    n = len(bwt)
    # Create a table of empty strings
    table = [""] * n

    # Repeat n times
    for _ in range(n):
        # Add the BWT as a new column in front of each row
        table = sorted([bwt[i] + table[i] for i in range(n)])

    # Find the row that ends with the end-of-string marker '$'
    for row in table:
        if row.endswith("$"):
            return row 
        
with open('bwt_data_thodoris', "r") as my_file:
    bwt_string = my_file.read().strip() 

original_seq = inverse_bwt(bwt_string)
print("The riginal protein sequence is:", original_seq)
```
